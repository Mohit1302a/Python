```python
#broadcasting in numpy  
import numpy as np
arr1=np.random.randint(1,88,size=(3,3))
print("Original Array:", arr1)
arr2=np.random.randint(1,88,size=(3,))
print("Array to be broadcasted:", arr2)

result=arr1+arr2
print("Result of broadcasting:", result)
```

```python
#linear algebra in numpy
arr3=np.random.randint(1,99,size=(3,3))
print("Original Array:", arr3)
determined_array=np.linalg.det(arr3)
print("Determinant of the array:", determined_array)
inverse_array=np.linalg.inv(arr3)
print("Inverse of the array:", inverse_array)
eigenvalues,eigenvectors=np.linalg.eig(arr3)
print("Eigenvalues of the array:", eigenvalues)
```
