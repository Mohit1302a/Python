```python
#practice numpy
import numpy as np
array1 = np.random.randint(1,10,size=(3,3))
print("Original Array:", array1)
#replace all the values greater than 5 with 0
array1[array1 > 5] = 0
#replace all the in 1 coulumns with 5
array1[:, 0] = 5
print("Array with first column replaced:", array1)  
#replace the daigonal values with 1
np.fill_diagonal(array1,1)
print("Array with diagonal values replaced:", array1)
#replace the values in the last row with 9
array1[-1, :] = 9
print("Array with last row replaced:", array1)

```

```python
#slicing the array
array1=np.arange(1,10).reshape(3,3)
print("Original Array:", array1)

#sub_array that contain element from 2 nd row and 2nd column
sub_array = array1[0:2, 0:2]
print("Sub-array:", sub_array)
```

```python
#extract elements from the border
array1 = np.arange(1,26).reshape(5,5)
print("Original Array:", array1)
border_elements=np.concatenate((array1[0,:],array1[-1,:],array1[1:-1,0],array1[1:-1,-1]))
print("Border Elements:", border_elements)
```

```python
#operation of two arrays
arr1=np.random.randint(1,10,size=(3,3))
arr2=np.random.randint(1,10,size=(3,3))
print("Array 1:", arr1)
print("Array 2:", arr2)
sum_array=arr1+arr2
print("Sum of two arrays:", sum_array)  
difference_array=arr1-arr2
print("Difference of two arrays:", difference_array)    
product_array=arr1*arr2
print("Product of two arrays:", product_array)
division_array=arr1/arr2
print("Division of two arrays:", division_array)


#row and cloumn wise sum
row_sum=np.sum(arr1, axis=1)
print("Row-wise sum:", row_sum )
column_sum=np.sum(arr1, axis=0) 
print("Column-wise sum:", column_sum)
```

```python
#statistical operations
from array import array


mean_value=np.mean(arr1)
print("Mean value:", mean_value)
median_value=np.median(arr1)
print("Median value:", median_value)
variance_value=np.var(arr1)
print("Variance value:", variance_value)
standard_deviation_value=np.std(arr1)
print("Standard Deviation value:", standard_deviation_value)   


normalized_array=(arr1-mean_value)/standard_deviation_value
print("Normalized Array:", normalized_array)
```
