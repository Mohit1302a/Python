```python
#bank account example
class BankAccount:
    def __init__(self,account_number,account_holder,balance):
        self.account_number=account_number
        self.account_holder=account_holder
        self.balance=balance
    def details(self):
        print(f"Account Number: {self.account_number}")
        print(f"Account Holder: {self.account_holder}")
        print(f"Balance: {self.balance}")
    def deposite(self,amount):
        self.balance+=amount
        print(f"new balance after deposite is{self.balance}")
    def withdraw(self,amount):
        if amount>self.balance:
            print("insufficient balance")
        else:
            self.balance-=amount
            print(f"new balance after withdraw is{self.balance}")
account=BankAccount("123456789","John Doe",1000)
account.details()
account.deposite(500)
account.withdraw(200)
```

```python
#inheritance in class
class Vehicle:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    
class Car(Vehicle):
    def __init__(self,make,model,year,color):
        super().__init__(make,model,year)
        self.color=color
    def details(self):
        print((self.make,self.model,self.year,self.color))
    def color_details(self):
        print(f"Color: {self.color}")
car=Car("Toyota","Camry",2020,"Red")
car.details()
car.color_details()
```

```python
#method overriding in class
class Vehicle:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    
    def details(self):
        print((self.make,self.model,self.year))
class Car(Vehicle):
    def __init__(self,make,model,year,color):
        super().__init__(make,model,year)
        self.color=color
    def details(self):
        print((self.make,self.model,self.year,self.color))
car=Car("Toyota","Camry",2020,"Red")
car.details()       
```

```python
#class composition
class Engine:
    def __init__(self,engine_type):
        self.engine_type=engine_type
    def details(self):
        print(f"Engine Type: {self.engine_type}")
class Car:
    def __init__(self,make,model,year,engine):
        self.make=make
        self.model=model
        self.year=year
        self.engine=engine
    def details(self):
        print((self.make,self.model,self.year))
        self.engine.details()
engine=Engine("V6")
car=Car("Toyota","Camry",2020,engine)
car.details()
```

```python
#class that counts the number of instances created
class Car:
    count=0
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
        Car.count+=1
    def details(self):
        print((self.make,self.model,self.year))
car1=Car("Toyota","Camry",2020)
car2=Car("Honda","Accord",2021)
print(f"Number of cars created: {Car.count},{car1.details()},{car2.details()}")   
```
