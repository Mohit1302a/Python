```python
#practice pandas 
import pandas as pd
import numpy as np
df=pd.DataFrame(np.random.randint(1,100,size=(4,4)),columns=['A','B','C','D'])
print("Original DataFrame:\n",df)
#set index to column 'A'
df.set_index('A',inplace=True)
print("DataFrame after setting index to column 'A':\n",df)

```

```python
#accesing the elemment
df=pd.DataFrame(np.random.randint(1,100,size=(4,4)),columns=['A','B','C','D'])
print("Original DataFrame:\n",df)   
elements=df.at[2,'B']
print("Element at row 2, column 'B':", elements)
```

```python
#operations on dataframe
df=pd.DataFrame(np.random.randint(1,100,size=(4,4)),columns=['W','X','Y','Z'])
print("Original DataFrame:\n",df)
df['V']=df['W']+df['X']
print("DataFrame after adding new column 'V':\n",df)
df['R']=df['Y']*df['Z']
print("DataFrame after adding new column 'R':\n",df)
row_sum=df.sum(axis=1)
print("Sum of each row:\n",row_sum)
column_sum=df.sum(axis=0)
print("Sum of each column:\n",column_sum)
```

```python
#data cleaning in pandas
df=pd.DataFrame(np.random.randint(1,100,size=(4,4)),columns=['A','B','C','D'])
print("Original DataFrame:\n",df)
#introducing missing values
df.loc[1,'B']=np.nan
df.loc[2,'C']=np.nan
print("DataFrame after introducing missing values:\n",df)
#fiilling missing values with mean
df_filled=df.fillna(df.mean(),inplace=True)
print("DataFrame after filling missing values with mean:\n",df)
#dropping rows with missing values
dropped_rows=df.dropna()
print("DataFrame after dropping rows with missing values:\n",dropped_rows)
```

```python
#merge the dataframes
df1=pd.DataFrame({'key':['A','B','C'],'data':[[1,2,3],[4,5,6],[7,8,9]]})
df2=pd.DataFrame({'key':['A','B','D'],'data':[[10,11,12],[13,14,15],[16,17,18]]})
print("DataFrame 1:\n",df1)
print("DataFrame 2:\n",df2)
merged_df=pd.merge(df1,df2,on='key',how='inner')
print("Merged DataFrame:\n",merged_df)
```

```python
#applying functions to dataframe
df=pd.DataFrame(np.random.randint(1,10,size=(4,4)),columns=['A','B','C','D'])
print("Original DataFrame:\n",df)
#applying lambda function to each element
df_squared=df.applymap(lambda x:x**2)
print("DataFrame after squaring each element:\n",df_squared)
#filtering rows based on condition using lambda function
filtered_df=df[df['A'].apply(lambda x:x>5)]
print("Filtered DataFrame with 'A' greater than 5:\n",filtered_df)
```
